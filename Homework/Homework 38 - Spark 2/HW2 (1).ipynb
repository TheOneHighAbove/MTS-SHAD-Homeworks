````python
users = pd.read_csv(os.path.join(data_path, "users.csv"))
items = pd.read_csv(os.path.join(data_path, "items.csv"))

users = users.sample(frac=0.1, random_state=42)

interactions = (
    pd.read_csv(os.path.join(data_path, "interactions.csv"), parse_dates=["last_watch_dt"])
    .rename(columns={'total_dur': rectools.Columns.Weight,
                     'last_watch_dt': rectools.Columns.Datetime})
)


interactions = interactions[interactions["user_id"].isin(users["user_id"])]


print(interactions.shape)
interactions.head(5)
````

````python
N_DAYS = 7

max_date = interactions['datetime'].max()
train = interactions[(interactions['datetime'] <= max_date - pd.Timedelta(days=N_DAYS))]
test = interactions[(interactions['datetime'] > max_date - pd.Timedelta(days=N_DAYS))]

catalog = train[rectools.Columns.Item].unique()

test_users = test[rectools.Columns.User].unique()
cold_users = set(test_users) - set(train[rectools.Columns.User])
test.drop(test[test[rectools.Columns.User].isin(cold_users)].index, inplace=True)
hot_users = test[rectools.Columns.User].unique()
print(test.shape[0])
print(test[rectools.Columns.User].nunique())

def scorer(map: float):
    print(f"Your MAP: {map}")
    UPPER_BOUND = 0.095
    LOWER_BOUND = 0.075
    score = int(min(max( (map - LOWER_BOUND) / (UPPER_BOUND - LOWER_BOUND), 0), 1) * 60)
    print(f"Your Score: {score}")
    return score
````

````python
def solution(train: pd.DataFrame, users: pd.DataFrame, items: pd.DataFrame):
    #  CODE BEGIN
    user_col = rectools.Columns.User
    item_col = rectools.Columns.Item
    date_col = rectools.Columns.Datetime

    train = train.copy()
    users_demo = users[[user_col, 'sex']].copy()
    users_demo['sex'] = users_demo['sex'].fillna('unknown')

    target_users = np.asarray(globals().get('hot_users', users[user_col].unique()))
    last_train_date = train[date_col].max()

    recent_global = train[train[date_col] > last_train_date - pd.Timedelta(days=7)]
    global_scores = (
        recent_global.groupby(item_col)[user_col]
        .nunique()
        .astype(float)
        .sort_values(ascending=False)
    )
    global_top = global_scores.head(700).index.to_list()
    global_scores = global_scores.to_dict()

    recent_by_sex = train[train[date_col] > last_train_date - pd.Timedelta(days=30)]
    recent_by_sex = recent_by_sex.merge(users_demo, on=user_col, how='left')
    recent_by_sex['sex'] = recent_by_sex['sex'].fillna('unknown')

    sex_tops = {}
    for sex_value, part in recent_by_sex.groupby('sex', dropna=False):
        scores = (
            part.groupby(item_col)[user_col]
            .nunique()
            .astype(float)
            .sort_values(ascending=False)
            .head(200)
        )
        sex_tops[sex_value] = (scores.index.to_list(), scores.to_dict())

    seen = train.groupby(user_col)[item_col].agg(set).to_dict()
    user_sex = users_demo.set_index(user_col)['sex'].to_dict()

    rows = []
    for user_id in target_users:
        watched = seen.get(user_id, set())
        sex_value = user_sex.get(user_id, 'unknown')
        sex_top, sex_scores = sex_tops.get(sex_value, ([], {}))

        candidates = []
        added = set()
        for item_id in global_top:
            candidates.append(item_id)
            added.add(item_id)
        for item_id in sex_top:
            if item_id not in added:
                candidates.append(item_id)
                added.add(item_id)

        scored = []
        for item_id in candidates:
            if item_id in watched:
                continue
            score = global_scores.get(item_id, 0.0) + sex_scores.get(item_id, 0.0)
            scored.append((score, item_id))

        scored.sort(reverse=True)
        for rank, (score, item_id) in enumerate(scored[:10], start=1):
            rows.append((user_id, item_id, score, rank))

    recos = pd.DataFrame(rows, columns=[user_col, item_col, rectools.Columns.Score, rectools.Columns.Rank])
    #  CODE END
    return recos
````

````python
%%time

recos = solution(train.copy(), users.copy(), items.copy())
scorer(rectools.metrics.MAP(10).calc(recos, test))
````
